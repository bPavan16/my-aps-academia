```python
import time
import tracemalloc

def subset_sum_recursive(arr, n, target):
    if target == 0:
        return True
    if n == 0:
        return False
    if arr[n - 1] > target:
        return subset_sum_recursive(arr, n - 1, target)
    return subset_sum_recursive(arr, n - 1, target) or subset_sum_recursive(arr, n - 1, target - arr[n - 1])

def subset_sum_dp(arr, n, target):
    dp = [[False] * (target + 1) for _ in range(n + 1)]
    for i in range(n + 1):
        dp[i][0] = True
    for i in range(1, n + 1):
        for j in range(1, target + 1):
            if arr[i - 1] > j:
                dp[i][j] = dp[i - 1][j]
            else:
                dp[i][j] = dp[i - 1][j] or dp[i - 1][j - arr[i - 1]]
    return dp[n][target]

def measure_performance(method, arr, n, target):
    tracemalloc.start()
    start_time = time.time()
    result = method(arr, n, target)
    end_time = time.time()
    memory_used = tracemalloc.get_traced_memory()[1] / 1024  # Convert to KB
    tracemalloc.stop()
    return result, (end_time - start_time) * 1000, memory_used  # Time in ms, Memory in KB

# Small Input
arr_small = [3, 34, 4, 12, 5, 2]
target_small = 9
n_small = len(arr_small)

# Large Input
arr_large = list(range(1, 61))  # 1 to 30
target_large = 300
n_large = len(arr_large)

# Running both methods on small input
rec_result_small, rec_time_small, rec_mem_small = measure_performance(subset_sum_recursive, arr_small, n_small, target_small)
dp_result_small, dp_time_small, dp_mem_small = measure_performance(subset_sum_dp, arr_small, n_small, target_small)

# Running both methods on large input
try:
    rec_result_large, rec_time_large, rec_mem_large = measure_performance(subset_sum_recursive, arr_large, n_large, target_large)
except RecursionError:
    rec_result_large, rec_time_large, rec_mem_large = "Stack Overflow", "Too High", "Too High"

dp_result_large, dp_time_large, dp_mem_large = measure_performance(subset_sum_dp, arr_large, n_large, target_large)

# Neatly formatted results
print("=" * 40)
print("         Subset Sum Performance        ")
print("=" * 40)

print("\n🔹 Small Input (arr = [3, 34, 4, 12, 5, 2], target = 9)")
print(f"\n    Recursive Approach")
print(f"       - Result: {'Found' if rec_result_small else 'Not Found'}")
print(f"       - Time Taken: {rec_time_small:.2f} ms")
print(f"       - Memory Used: {rec_mem_small:.2f} KB")

print(f"\n    Dynamic Programming (DP) Approach")
print(f"       - Result: {'Found' if dp_result_small else 'Not Found'}")
print(f"       - Time Taken: {dp_time_small:.2f} ms")
print(f"       - Memory Used: {dp_mem_small:.2f} KB")

print("-" * 40)
print("\n🔹 Large Input (arr = [1, 2, ..., 60], target = 300)")
print(f"\n    Recursive Approach")
print(f"       - Result: {rec_result_large}")
print(f"       - Time Taken: {rec_time_large}")
print(f"       - Memory Used: {rec_mem_large}")

print(f"\n    Dynamic Programming (DP) Approach")
print(f"       - Result: {'Found' if dp_result_large else 'Not Found'}")
print(f"       - Time Taken: {dp_time_large:.2f} ms")
print(f"       - Memory Used: {dp_mem_large:.2f} KB")

print("=" * 40)
print("DP is significantly better for large inputs!")
print("=" * 40)
```
