# What do you mean by descriptives

"Descriptives, no code" means a request for summary statistics or descriptive information about data *without* requiring any code to be executed.  Essentially, you're asking for a human-readable summary of the data's characteristics, rather than a programmatically generated output.  

For example, if you have a dataset of ages, you might ask for descriptives like:

* **Mean:** Average age
* **Median:** Middle age
* **Standard deviation:** Measure of the spread of the ages
* **Minimum and maximum:** Youngest and oldest ages
* **Number of observations:** How many ages are in the dataset

You could also ask for descriptive information about the distribution of the data, such as whether it is skewed or normally distributed.  This descriptive information is often done visually using histograms, box plots or other data visualizations.  However, the request of "descriptives no code" implies a textual description, not visual representations requiring code.


```python
# prompt: Generate 100 Random number and plot a distribution Mark a mean median mode on the graph

import random
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

# Generate 100 random numbers
random_numbers = [random.randint(1, 100) for _ in range(100)]

# Calculate mean, median, and mode
mean = np.mean(random_numbers)
median = np.median(random_numbers)
# Access the mode value using .mode and convert to int if necessary
mode = stats.mode(random_numbers).mode

# Create the histogram
plt.hist(random_numbers, bins=10, edgecolor='black')

# Add mean, median, and mode lines
plt.axvline(mean, color='red', linestyle='dashed', linewidth=1, label=f'Mean: {mean:.2f}')
plt.axvline(median, color='green', linestyle='dashed', linewidth=1, label=f'Median: {median:.2f}')
plt.axvline(mode, color='blue', linestyle='dashed', linewidth=1, label=f'Mode: {mode:.2f}')

# Customize the plot
plt.xlabel('Random Numbers')
plt.ylabel('Frequency')
plt.title('Distribution of 100 Random Numbers')
plt.legend()

# Show the plot
plt.show()
```

```python
# prompt: # prompt: Generate 100 Random number whose mean is 50 sd = 20 and plot a distribution Mark a mean median mode on the graph
# find mode using KDE

import random
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
from scipy.stats import gaussian_kde

# Generate 100 random numbers with mean 50 and standard deviation 20
mu = 50
sigma = 20
random_numbers = np.random.normal(mu, sigma, 100)

# Calculate mean and median
mean = np.mean(random_numbers)
median = np.median(random_numbers)

# Calculate mode using KDE
density = gaussian_kde(random_numbers)
x = np.linspace(min(random_numbers), max(random_numbers), 1000)
density_values = density(x)
mode = x[np.argmax(density_values)]


# Create the histogram
plt.hist(random_numbers, bins=10, edgecolor='black', alpha=0.6, label='Histogram')

# Plot the KDE
plt.plot(x, density_values * len(random_numbers) * (x[1]-x[0]), label='KDE', color='orange')

# Add mean, median, and mode lines
plt.axvline(mean, color='red', linestyle='dashed', linewidth=1, label=f'Mean: {mean:.2f}')
plt.axvline(median, color='green', linestyle='dashed', linewidth=1, label=f'Median: {median:.2f}')
plt.axvline(mode, color='blue', linestyle='dashed', linewidth=1, label=f'Mode: {mode:.2f}')

# Customize the plot
plt.xlabel('Random Numbers')
plt.ylabel('Frequency/Density')
plt.title('Distribution of 100 Random Numbers (Mean=50, SD=20)')
plt.legend()

# Show the plot
plt.show()
```

```python
# prompt: # prompt: for the same data whose mean = 50 and SD=20 genrate a Graph which is Right skewed

import numpy as np
import matplotlib.pyplot as plt

# Generate right-skewed data
mu = 50
sigma = 20
# Use a gamma distribution to create right-skewed data
# Adjust 'a' and 'scale' to control skewness, mean, and standard deviation
a = 2  # Shape parameter
scale = sigma / np.sqrt(a)  # Scale parameter to match the desired standard deviation

# Generate data to have the same mean
random_numbers = (np.random.gamma(a, scale, 100) - np.random.gamma(a, scale, 100).mean() + mu)



# Calculate mean, median, and mode (approximate using KDE)
mean = np.mean(random_numbers)
median = np.median(random_numbers)

from scipy.stats import gaussian_kde

density = gaussian_kde(random_numbers)
x = np.linspace(min(random_numbers), max(random_numbers), 1000)
density_values = density(x)
mode = x[np.argmax(density_values)]

# Create the histogram
plt.hist(random_numbers, bins=10, edgecolor='black', alpha=0.6, label='Histogram')

# Plot the KDE
plt.plot(x, density_values * len(random_numbers) * (x[1] - x[0]), label='KDE', color='orange')

# Add mean, median, and mode lines
plt.axvline(mean, color='red', linestyle='dashed', linewidth=1, label=f'Mean: {mean:.2f}')
plt.axvline(median, color='green', linestyle='dashed', linewidth=1, label=f'Median: {median:.2f}')
plt.axvline(mode, color='blue', linestyle='dashed', linewidth=1, label=f'Mode: {mode:.2f}')

# Customize the plot
plt.xlabel('Random Numbers')
plt.ylabel('Frequency/Density')
plt.title('Right-Skewed Distribution (Mean=50, SD=20)')
plt.legend()

# Show the plot
plt.show()
```

```python
# prompt: for the same data whose mean = 50 and SD=20 genrate a Graph which is left skewed

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skewnorm

# Parameters for the skewed normal distribution
mean = 50
std_dev = 20
skewness = -5  # Negative skewness for left skew

# Generate random numbers from the skewed normal distribution
random_numbers = skewnorm.rvs(a=skewness, loc=mean, scale=std_dev, size=100)

# Calculate mean, median, and mode
mean = np.mean(random_numbers)
median = np.median(random_numbers)
mode = random_numbers[np.argmax(np.histogram(random_numbers, bins=10)[0])]

# Create the histogram
plt.hist(random_numbers, bins=10, edgecolor='black', alpha=0.6, label='Histogram')


# Add mean, median, and mode lines
plt.axvline(mean, color='red', linestyle='dashed', linewidth=1, label=f'Mean: {mean:.2f}')
plt.axvline(median, color='green', linestyle='dashed', linewidth=1, label=f'Median: {median:.2f}')
plt.axvline(mode, color='blue', linestyle='dashed', linewidth=1, label=f'Mode: {mode:.2f}')

# Customize the plot
plt.xlabel('Random Numbers')
plt.ylabel('Frequency')
plt.title('Left-Skewed Distribution (Mean=50, SD=20)')
plt.legend()

# Show the plot
plt.show()
```

```python
# prompt: generate a graph where in company A mean salary 100 employes is 50 thousand and the salary ranges from 10 thousand to 2 lakhs . In company B the mean salary is 50000 AND the salary ranges from 30,000 to 70,000 draw the graph and show mean median mode overlay one grpah on top of one another

import matplotlib.pyplot as plt
import numpy as np

# Company A
mean_a = 50000
min_a = 10000
max_a = 200000
employees_a = 100

# Generate salary data for Company A (simulated)
np.random.seed(0)
salary_a = np.random.uniform(min_a, max_a, employees_a)


# Company B
mean_b = 50000
min_b = 30000
max_b = 70000
employees_b = 100 # Assuming same number of employees for comparison

# Generate salary data for Company B (simulated)
salary_b = np.random.uniform(min_b, max_b, employees_b)

# Calculate median and mode for both companies (approximation for mode)

median_a = np.median(salary_a)
mode_a = np.argmax(np.histogram(salary_a, bins=10)[0])


median_b = np.median(salary_b)
mode_b = np.argmax(np.histogram(salary_b, bins=10)[0])


# Plotting

plt.figure(figsize=(12, 6))

plt.hist(salary_a, bins=20, alpha=0.5, label='Company A', color='skyblue', density=True)
plt.hist(salary_b, bins=20, alpha=0.5, label='Company B', color='salmon', density=True)


plt.axvline(mean_a, color='blue', linestyle='dashed', linewidth=1, label=f'Company A Mean: {mean_a}')
plt.axvline(median_a, color='blue', linestyle='dotted', linewidth=1, label=f'Company A Median: {median_a:.0f}')
plt.axvline(mode_a * (max_a - min_a)/10 + min_a, color='blue', linestyle='dashdot', linewidth=1, label=f'Company A Mode (approx): {mode_a * (max_a - min_a)/10 + min_a:.0f}')

plt.axvline(mean_b, color='red', linestyle='dashed', linewidth=1, label=f'Company B Mean: {mean_b}')
plt.axvline(median_b, color='red', linestyle='dotted', linewidth=1, label=f'Company B Median: {median_b:.0f}')
plt.axvline(mode_b* (max_b - min_b)/10 + min_b, color='red', linestyle='dashdot', linewidth=1, label=f'Company B Mode (approx): {mode_b * (max_b - min_b)/10 + min_b:.0f}')

plt.xlabel('Salary')
plt.ylabel('Density')
plt.title('Salary Distribution in Company A and B')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# prompt: Generate 100 Random numbers and draw a box plot introduce few outliers mark the mean and median

import matplotlib.pyplot as plt
import numpy as np

# Generate 100 random data points with a few outliers
data = np.random.normal(loc=50, scale=10, size=90)
outliers = np.random.uniform(low=100, high=150, size=10)
data = np.concatenate((data, outliers))

# Calculate mean and median
mean = np.mean(data)
median = np.median(data)

# Create the box plot
plt.figure(figsize=(8, 6))  # Adjust figure size for better visualization
plt.boxplot(data, vert=False, patch_artist=True, showmeans=True, meanline=True,
            medianprops={'color': 'green', 'linewidth': 2},
            meanprops={'marker': 'o', 'markerfacecolor': 'red', 'markeredgecolor': 'black', 'markersize': 8},
            boxprops={'facecolor': 'lightblue'})

# Add mean and median lines
plt.axvline(x=mean, color='red', linestyle='dashed', linewidth=1.5, label=f'Mean: {mean:.2f}')
plt.axvline(x=median, color='green', linestyle='dashed', linewidth=1.5, label=f'Median: {median:.2f}')

# Customize the plot
plt.title('Box Plot of Random Data with Outliers', fontsize=14)
plt.xlabel('Value', fontsize=12)
plt.yticks([]) # Remove the y-axis ticks as we only have one box plot
plt.legend()
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout() # Adjust layout to prevent labels from overlapping

# Show the plot
plt.show()
```
